# 1. Método de Congruencias Lineales (LCG) - Implementación Completa

## Fórmula Matemática

$$X_{n+1} = (a \times X_n + c) \bmod m$$

Donde:
- **a**: Multiplicador = 1103515245 (estándar glibc)
- **c**: Incremento = 12345 (estándar glibc)
- **m**: Módulo = 2^32 = 4,294,967,296
- **X_n**: Semilla actual

```python
# Importar el generador LCG
import sys
sys.path.insert(0, './src')

from duck_hunt.prng_lcg import LCG

# Parámetros del LCG (32-bits)
a = 1103515245
c = 12345
m = 2**32
Xo = 12345  # Semilla

print("=" * 70)
print("GENERADOR DE CONGRUENCIAS LINEALES - PARÁMETROS (32-BITS)")
print("=" * 70)
print(f"a (Multiplicador):  {a:>15,}")
print(f"c (Incremento):     {c:>15,}")
print(f"m (Módulo):         {m:>15,}")
print(f"Xo (Semilla):       {Xo:>15,}")
print("=" * 70)
```

## Generación de Números Decimales [0.0, 1.0)

```python
# Instanciar el generador
prng = LCG(Xo)

print("\nGenerando 10 números decimales:")
print("-" * 70)

for i in range(10):
    numero = prng.next_float()
    print(f"X_{i+1:2d} = {numero:.10f}")
```

# 2. Aplicación: Simulación de Dado (1-6) para Duck Hunt

Uno de los usos del PRNG en Duck Hunt es seleccionar mapas aleatorios.

**Mapeo de rango [0,1) → [1,6]:**

$$X_{mapa} = \lfloor Pendiente \times X_n + Intercepto \rfloor$$

```python
# Reiniciar con semilla
prng.set_seed(Xo)

# Parámetros del rango (límites para selección de mapa)
Lim_1 = 1
Lim_2 = 6

Pendiente = (Lim_2 + 1 - Lim_1) / (1 - 0)  # = 6
Intercepto = Lim_1  # = 1

print("\nSimulando 20 selecciones de mapa (1-6):")
print("-" * 70)
print(f"Pendiente:  {Pendiente}")
print(f"Intercepto: {Intercepto}")
print("-" * 70)

Num_samples = 20
Lista_mapas = []

for i in range(Num_samples):
    numero_rango = prng.randint(Lim_1, Lim_2)
    Lista_mapas.append(numero_rango)
    print(f"Selección {i+1:2d}: Mapa {numero_rango}")

print(f"\nSecuencia de mapas generada: {Lista_mapas}")
```

# 3. Mapeo a Mapas Reales del Juego

```python
# Mapas disponibles en Duck Hunt
mapas_disponibles = [
    "Bosque Oscuro",
    "Montaña Nevada", 
    "Pantano Misterioso",
    "Playa Tropical",
    "Castillo Antiguo",
    "Cueva Subterránea"
]

# Reiniciar generador
prng.set_seed(Xo)

print("\nMapeos de selecciones a mapas reales:")
print("-" * 70)

for i, num_mapa in enumerate(Lista_mapas, 1):
    indice = (num_mapa - 1) % len(mapas_disponibles)
    mapa_seleccionado = mapas_disponibles[indice]
    print(f"Ronda {i:2d}: Selección={num_mapa} → Mapa: {mapa_seleccionado}")
```

# 4. Reproducibilidad: Misma Semilla = Misma Secuencia

```python
print("\nVERIFICACIÓN DE REPRODUCIBILIDAD")
print("=" * 70)

# Generar secuencia 1
prng1 = LCG(99999)
secuencia1 = [prng1.next_float() for _ in range(5)]

# Generar secuencia 2 con MISMA semilla
prng2 = LCG(99999)
secuencia2 = [prng2.next_float() for _ in range(5)]

print("\nSecuencia 1 (Semilla=99999):")
for i, val in enumerate(secuencia1, 1):
    print(f"  X_{i} = {val:.10f}")

print("\nSecuencia 2 (Semilla=99999):")
for i, val in enumerate(secuencia2, 1):
    print(f"  X_{i} = {val:.10f}")

if secuencia1 == secuencia2:
    print("\n✓ REPRODUCIBILIDAD VERIFICADA")
    print("  Las secuencias son idénticas.")
    print("  Esto permite depurar y validar el juego.")
else:
    print("\n✗ ERROR: Las secuencias no coinciden")
```

# 6. Ejecución del Juego Duck Hunt

## Información del Juego

El juego Duck Hunt utiliza el generador LCG de 32-bits para:

1. **Selección aleatoria de mapas** - Sin usar semilla (cada partida es diferente)
2. **Lógica de juego reproducible** - Con semilla controlada (para QA/testing)
3. **Efectos visuales** - Sangre y partículas
4. **Audio** - Música y efectos de sonido
5. **Sistemas de scoring** - Rankings persistentes

```python
print("\nEJECUTANDO DUCK HUNT CON PRNG LCG 32-BITS")
print("=" * 70)

# Simular estadísticas del juego
prng_game = LCG(42)  # Semilla fija para reproducibilidad en esta demo

print("\nRonda 1: Iniciando partida...")
print("-" * 70)

# Seleccionar mapa aleatorio
num_mapa = prng_game.randint(1, 6)
mapas = ["Bosque", "Montaña", "Pantano", "Playa", "Castillo", "Cueva"]
mapa_seleccionado = mapas[num_mapa - 1]

print(f"Mapa seleccionado: {mapa_seleccionado}")

# Simular cantidad de enemigos
num_enemigos = prng_game.randint(5, 10)
print(f"Cantidad de enemigos: {num_enemigos}")

# Simular disparos y aciertos
aciertos = 0
disparos = 0

print("\nSimulando tiradas...")
for i in range(num_enemigos):
    # Probabilidad de acertar: 60%
    if prng_game.next_float() < 0.6:
        aciertos += 1
        resultado = "¡ACERTASTE! 💥"
    else:
        resultado = "Fallaste..."
    disparos += 1
    print(f"  Disparo {disparos}: {resultado}")

# Calcular puntuación
puntuacion = aciertos * 100

print("\n" + "-" * 70)
print(f"\nResultados finales:")
print(f"  Aciertos: {aciertos}/{disparos}")
print(f"  Precisión: {(aciertos/disparos)*100:.1f}%")
print(f"  Puntuación: {puntuacion} puntos")
print("\n✓ Partida completada con PRNG LCG 32-bits")
```
